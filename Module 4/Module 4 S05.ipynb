```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('regularization_comparison_s26.csv')
print("imported")
df.head()
```

```python
# Model comparison data
data = {
    'Model': [
        'Unregularized LR',
        'Ridge (L2)',
        'Lasso (L1)',
        'ElasticNet'
    ],
    'R2': [0.923820, 0.917973, 0.899018, 0.899018],
    'RMSE': [21.826064, 22.648182, 25.129109, 25.129109],
    'MAE': [15.009205, 15.157836, 17.207660, 17.207660]
}

df = pd.DataFrame(data)

# Display the data
print("Model Performance Comparison:")
print(df.to_string(index=False))

# Create three comparison plots
fig, ax = plt.subplots(1, 3, figsize=(18, 6))

# 1. R² Comparison (higher is better)
ax[0].bar(df['Model'], df['R2'])
ax[0].set_title('R² Comparison')
ax[0].set_ylabel('R² Score')
ax[0].set_ylim(0, 1)
ax[0].tick_params(axis='x', rotation=45)

# 2. RMSE Comparison (lower is better)
ax[1].bar(df['Model'], df['RMSE'])
ax[1].set_title('RMSE Comparison')
ax[1].set_ylabel('RMSE')
ax[1].tick_params(axis='x', rotation=45)

# 3. MAE Comparison (lower is better)
ax[2].bar(df['Model'], df['MAE'])
ax[2].set_title('MAE Comparison')
ax[2].set_ylabel('MAE')
ax[2].tick_params(axis='x', rotation=45)

plt.suptitle('Regression Model Diagnostics Comparison')
plt.tight_layout()
plt.show()
```
