```python
# ============================================================
# 1. Import the Matplotlib library
# ============================================================

import matplotlib.pyplot as plt
# import      -> Loads a Python library.
# matplotlib  -> A library used for creating graphs and plots.
# pyplot      -> A module that provides plotting functions.
# as plt      -> Creates the shortcut name 'plt'.
```

```python



# ============================================================
# 2. Store the data
# ============================================================

elements = ['C', 'Si', 'Ge', 'Sn', 'Pb']
# elements -> List containing the symbols of Group 14 elements.

atomic_numbers = [6, 14, 32, 50, 82]
# atomic_numbers -> X-axis values (Atomic Number of each element).

bandgaps = [5.47, 1.11, 0.67, 0.10, 0.00]
# bandgaps -> Y-axis values (Bandgap Energy in electron volts).


# ============================================================
# 3. Create a figure
# ============================================================

plt.figure(figsize=(8, 5))
# figure()      -> Creates a new plotting window.
# figsize=(8,5) -> Width = 8 inches, Height = 5 inches.


# ============================================================
# 4. Plot the graph
# ============================================================

plt.plot(
    atomic_numbers,      # X-axis data
    bandgaps,            # Y-axis data
    marker='o',          # Draw circular markers
    color='blue',        # Line color is blue
    linestyle='-',       # Solid line
    linewidth=2          # Line thickness
)


# ============================================================
# 5. Add graph title
# ============================================================

plt.title(
    'Bandgap vs. Atomic Number (Group 14 Elements)',
    fontsize=14
)
# title()      -> Adds title to the graph.
# fontsize=14  -> Title font size.


# ============================================================
# 6. Label the X-axis
# ============================================================

plt.xlabel('Atomic Number (Z)', fontsize=12)
# xlabel()     -> Labels the horizontal axis.
# fontsize=12  -> Font size of X-axis label.


# ============================================================
# 7. Label the Y-axis
# ============================================================

plt.ylabel('Bandgap Energy (eV)', fontsize=12)
# ylabel()     -> Labels the vertical axis.
# fontsize=12  -> Font size of Y-axis label.


# ============================================================
# 8. Display element names beside each point
# ============================================================

for i, element in enumerate(elements):
    # for           -> Starts a loop.
    # i             -> Index number (0,1,2,3,4).
    # element       -> Current element name (C, Si, Ge, Sn, Pb).
    # enumerate()   -> Returns both index and value.

    plt.text(
        atomic_numbers[i],   # X-coordinate
        bandgaps[i],         # Y-coordinate
        element,             # Text to display
        fontsize=10,         # Text size
        ha='left',           # Horizontal alignment
        va='bottom'          # Vertical alignment
    )


# ============================================================
# 9. Add grid
# ============================================================

plt.grid(
    True,              # Turn grid ON
    linestyle='--',    # Dashed grid lines
    alpha=0.7          # Transparency (0 = invisible, 1 = fully visible)
)


# ============================================================
# 10. Display the graph
# ============================================================

plt.show()
# show() -> Displays the completed graph.
```
