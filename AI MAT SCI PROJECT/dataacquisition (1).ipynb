```python
# ============================================================
# STEP 3: Enter your Materials Project API key
# ============================================================

API_KEY = "yG9MO3EiaegBkJOVJauCFubkj8QJgeGr"
```

```python
# ============================================================
# STEP 4: Retrieve 10,000 materials from Materials Project
# ============================================================

from mp_api.client import MPRester

print("Connecting to Materials Project API...")

with MPRester(API_KEY) as mpr:

    docs = mpr.materials.summary.search(
        fields=[
            "material_id",
            "formula_pretty",
            "symmetry",
            "structure"
        ],
        num_chunks=100,
        chunk_size=100
    )

print("Data retrieved successfully!")
print("Number of materials retrieved:", len(docs))
```

```python
# ============================================================
# STEP 5: Extract chemical compositions and crystal structures
# ============================================================

data = []

for doc in docs:

    # Extract material ID
    material_id = str(doc.material_id)

    # Extract chemical formula
    chemical_formula = doc.formula_pretty

    # Extract crystal system
    if doc.symmetry and doc.symmetry.crystal_system:
        crystal_system = doc.symmetry.crystal_system.value
    else:
        crystal_system = None

    # Extract space group
    if doc.symmetry:
        space_group = doc.symmetry.symbol
    else:
        space_group = None

    # Convert full crystal structure into JSON text
    if doc.structure:
        full_structure = json.dumps(doc.structure.as_dict())
    else:
        full_structure = None

    # Store all extracted information
    data.append({
        "material_id": material_id,
        "chemical_formula": chemical_formula,
        "crystal_system": crystal_system,
        "space_group": space_group,
        "full_structure": full_structure
    })
# ============================================================
# STEP 6: Create a Pandas DataFrame
# ============================================================

df = pd.DataFrame(data)

print("\nExtracted Data:")
display(df)

# ============================================================
# STEP 7: Save the complete dataset as CSV
# ============================================================

full_csv_file = "materials_full_structure_data.csv"

df.to_csv(full_csv_file, index=False)

print("\nFull structure CSV file created successfully!")
print("File name:", full_csv_file)

# ============================================================
# STEP 8: Save a simplified CSV for machine learning
# ============================================================

ml_df = df[
    [
        "material_id",
        "chemical_formula",
        "crystal_system",
        "space_group"
    ]
]

ml_csv_file = "materials_composition_crystal_data.csv"

ml_df.to_csv(ml_csv_file, index=False)

print("\nComposition and crystal-system CSV created successfully!")
print("File name:", ml_csv_file)

# ============================================================
# STEP 9: Display the first five records
# ============================================================

print("\nFirst five records:")
display(ml_df.head())
```
